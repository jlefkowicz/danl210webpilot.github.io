# Spotify Analysis

```python
import pandas as pd
from google.colab import data_table
data_table.enable_dataframe_formatter()
spotify = pd.read_csv('https://bcdanl.github.io/data/spotify_all.csv')
spotify
```

## Favorite Artist
* my favorite artist is dire straits
* I will be analyzing their info from spotify

```python
spotify_ds = spotify[spotify['artist_name'] == 'Dire Straits']
spotify_ds
```

### Playlists
* Dire Straits appear in 26 different playlists in this dataframe
* they appear the most in the "Classic Rock" data frame
* on average dire straits are added at the 54th spot on people's playlists

```python
spotify_ds['playlist_name'].nunique()
```

```python
spotify['pos'].mean()
```

```python
playlist_count = spotify_ds['playlist_name'].value_counts()
playlist_count
```

### Albums
* three dire straits albums appear in this dataframe
* Brothers in arms being the most popular
* Making movies being the least popular

```python
album_count = spotify_ds['album_name'].value_counts()
album_count
```

### Songs
* Sultans of swing is the most popular
